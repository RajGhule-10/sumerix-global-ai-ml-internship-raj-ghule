# Task 2: Feature Engineering, Model Optimization & Performance Comparison

This project compares multiple machine learning regression models on the California Housing Dataset. The objective is to understand how preprocessing and model selection impact prediction performance.

## Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor

from sklearn.metrics import mean_squared_error, r2_score
```

## Load Dataset

```python
# Loading California Housing Dataset

data = fetch_california_housing(as_frame=True)

df = pd.concat(
    [data.data, data.target.rename("HousePrice")],
    axis=1
)

df.head()
```

## Dataset Overview

```python
print("Dataset Shape:", df.shape)

df.info()
```

```python
df.describe()
```

## Separate Features and Target Variable

```python
# Features

X = df.drop("HousePrice", axis=1)

# Target

y = df["HousePrice"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

## Feature Scaling

```python
# Standardizing features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Feature scaling completed.")
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## Train Multiple Models

```python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Decision Tree": DecisionTreeRegressor(
        max_depth=5,
        random_state=42
    )
}
```

## Model Evaluation and Comparison

```python
results = {}

for name, model in models.items():

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    results[name] = {
        "RMSE": rmse,
        "R2 Score": r2
    }

results_df = pd.DataFrame(results).T

results_df
```

## Best Performing Model

```python
best_model_name = results_df["R2 Score"].idxmax()

print(
    f"Best Performing Model: {best_model_name}"
)

results_df.sort_values(
    by="R2 Score",
    ascending=False
)
```

## Performance Comparison Visualization

```python
results_df["R2 Score"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Model Comparison using R2 Score")

plt.ylabel("R2 Score")

plt.show()
```

## Visual Performance Validation

```python
best_model = models[best_model_name]

y_pred = best_model.predict(X_test)

plt.figure(figsize=(8,6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.4
)

plt.xlabel("Actual House Prices")
plt.ylabel("Predicted House Prices")

plt.title(
    f"Actual vs Predicted ({best_model_name})"
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="red"
)

plt.show()
```

## Conclusion

Three regression models were trained and evaluated on the California Housing Dataset.

Feature scaling was applied before training to ensure fair learning across features.

Model performance was compared using RMSE and R² Score. The best-performing model was selected based on evaluation metrics.

This task demonstrated how machine learning engineers compare multiple algorithms and choose the most suitable model for a prediction problem.
